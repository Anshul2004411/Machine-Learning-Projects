```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['label']=encoder.fit_transform(df.label)
```

```python
x=df.drop(['label'],axis=1)
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc= StandardScaler()
X_train_sc=sc.fit_transform(x_train)
X_test_sc=sc.transform(x_test)
```

```python
import keras
from keras.models import Sequential
from keras.layers import Dense, Dropout
```

```python
classifier = Sequential()
```

```python
classifier.add(Dense(units=64,
                     activation='relu',
                     input_shape=(X_train_sc.shape[1],)))

```

```python
classifier.add(Dropout(rate=0.1))
```

```python
classifier.add(Dense(units=64,
                     activation='relu',))
classifier.add(Dropout(rate=0.1))
```

```python
classifier.add(Dense(units=22,activation='softmax'))
```

```python
classifier.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
classifier.fit(
    X_train_sc,
    y_train,
    batch_size=32,
    epochs=200
)
```

```python
import pandas as pd
import numpy as np

def recomend(a,b,c,d,e,f,g):

    input_data = pd.DataFrame(
        [[a,b,c,d,e,f,g]]
    )

    scaled_input = sc.transform(input_data)

    prediction = classifier.predict(scaled_input, verbose=0)

    predicted_class_idx = np.argmax(prediction)

    recomend_crop_name = encoder.inverse_transform([predicted_class_idx])[0]

    confidence = prediction[0][predicted_class_idx] * 100

    return recomend_crop_name, confidence
```

```python
sample_crop, sample_conf = recomend(
    60,55,44,23.004459,82.320763,7.840207,263.964248
)

print(sample_crop)
print(sample_conf)
```
