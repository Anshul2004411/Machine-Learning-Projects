```python
#Facial features analyzer
import cv2

# Load cascades using OpenCV's built-in path
face_cascade = cv2.CascadeClassifier(r"C:/Users/anshu/Downloads/haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier(r"C:/Users/anshu/Downloads/haarcascade_eye.xml")

# Capture video from webcam
cap = cv2.VideoCapture(0)

while True:
    ret, img = cap.read()
    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = face_cascade.detectMultiScale(gray, 1.1, 4)

    # Draw rectangles around faces and eyes
    for (x, y, w, h) in faces:
        cv2.rectangle(img, (x, y), (x + w, y + h), (255, 0, 0), 2)
        roi_gray = gray[y:y + h, x:x + w]
        roi_color = img[y:y + h, x:x + w]

        eyes = eye_cascade.detectMultiScale(roi_gray)
        for (ex, ey, ew, eh) in eyes:
            cv2.rectangle(roi_color, (ex, ey), (ex + ew, ey + eh), (0, 255, 0), 2)
    
    #Display
    cv2.imshow('img',img)
    #stop if escape key is pressed
    if cv2.waitKey(1)&0xFF==ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
#calculation faces in a frame
import cv2
import numpy as np
import dlib

cap=cv2.VideoCapture(0)
detector=dlib.get_frontal_face_detector()

while True:
    ret, frame = cap.read()
    frame=cv2.flip(frame,1)
    gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    faces=detector(gray)
    i=0
    for face in faces:
        x,y=face.left(),face.top()
        x1,y1=face.right(),face.bottom()
        cv2.rectangle(frame,(x,y),(x1,y1),(0,255,0),2)
        i=i+1
        cv2.putText(frame, "Face num"+str(i),(x-10,y-10),cv2.FONT_HERSHEY_SIMPLEX,0.7,(100,200,255),3)
        print(face,i)
    cv2.imshow('frame',frame)
    if cv2.waitKey(1)&0xFF==ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
#Age Analyzer
import cv2
from deepface import DeepFace

#load Harcascad file of frontal face

face_cascade = cv2.CascadeClassifier(r"C:\Users\anshu\Downloads\haarcascade_frontalface_default.xml")

#start webcam

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30,30)
    )

    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    for (x,y,w,h) in faces:

        face_roi = rgb_frame[y:y+h, x:x+w]

        result = DeepFace.analyze(
            face_roi,
            actions=['age'],
            enforce_detection=False
        )

        if isinstance(result, list):
            age = result[0]['age']
        else:
            age = result['age']

        cv2.rectangle(
            frame,
            (x,y),
            (x+w,y+h),
            (0,0,255),
            2
        )

        cv2.putText(
            frame,
            f"Age: {int(age)}",
            (x,y-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0,0,255),
            2
        )

    cv2.imshow("Age Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```

```python
#Recording
import cv2

cap=cv2.VideoCapture(0)

width=int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height=int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

writer=cv2.VideoWriter()

while True:
    ret,frame=cap.read()
    
    writer.write(frame)
    
    cv2.imshow('Recording',frame)
    
    if cv2.waitKey(1) & 0xFF == 27:
        break
        
cap.release()
writer.release()
cv2.destroyAllWindows()
```

```python
#car count
import cv2
cascade_src="C:/Users/anshu/Downloads/cars.xml"
video_src="C:/Users/anshu/Downloads/CarVideo.mp4"

cap=cv2.VideoCapture(video_src)
fgbg=cv2.createBackgroundSubtractorMOG2()
car_cascade=cv2.CascadeClassifier(cascade_src)

while True:
    ret,img=cap.read()     
    fgbg.apply(img)
    if(type(img)==type(None)):
        break
    gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
    cars = car_cascade.detectMultiScale(gray, 2.5, 1)
    
    for(x,y,w,h) in cars:
        cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,255),2)
    
    cv2.imshow('Video', img)
    if cv2.waitKey(80)==27:
        break
        
cap.release()
cv2.destroyAllWindows()
```

```python
#Walking person photo
import cv2
import imutils
hog=cv2.HOGDescriptor()
hog.setSVMDetector(cv2.HOGDescriptor_getDefaultPeopleDetector())
image=cv2.imread(r"C:\Users\anshu\OneDrive\Desktop\img.jpg")
image=imutils.resize(image,width=min(500,image.shape[1]))
(regions,_)=hog.detectMultiScale(image,winStride=(4,4),padding=(4,4),scale=1.05)

for(x,y,w,h) in regions:
    cv2.rectangle(image,(x,y),(x+w,y+h),(500,200,0),2)
cv2.imshow("Walking",image)
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
#WASTE COLLECTION
import cv2
import cvzone
from cvzone.ClassificationModule import Classifier
import os

cap = cv2.VideoCapture(0)
classifier=Classifier("C:/Users/anshu/Downloads/Resources (2)/Resources/Resources/Model/keras_model.h5","C:/Users/anshu/Downloads/Resources (2)/Resources/Resources/Model/labels.txt")
imgArrow =cv2.imread("C:/Users/anshu/Downloads/Resources (2)/Resources/Resources/arrow.png",cv2.IMREAD_UNCHANGED)
classIDBin = 0
#import all the waste images
imgWasteList = []
pathFolderWaste = "C:/Users/anshu/Downloads/Resources (2)/Resources/Resources/Waste"
pathList = os.listdir(pathFolderWaste)
for path in pathList:
    imgWasteList.append(cv2.imread(os.path.join(pathFolderWaste, path),cv2.IMREAD_UNCHANGED))

#import all the waste images
imgBinsList =[]
pathFolderBins = "C:/Users/anshu/Downloads/Resources (2)/Resources/Resources/Bins"
pathList=os.listdir(pathFolderBins)
for path in pathList:
    imgBinsList.append(cv2.imread(os.path.join(pathFolderBins,path),cv2.IMREAD_UNCHANGED))

#0 = Recyclable
#1 =Hazardous
#2=Food
#3 = Residual

classDic={ 0:None,
           1:0,2:0,
           3:3,4:3,
           5:1,6:1,
           7:2,8:2}
while True:
    success, img = cap.read()
    if not success:
        break

    imgResize = cv2.resize(img, (454, 340))
    imgBackground = cv2.imread("C:/Users/anshu/Downloads/Resources (2)/Resources/Resources/background.png")

    prediction = classifier.getPrediction(img)
    classID = prediction[1]
    print("ClassID:", classID)

    if classID != 0:
        imgBackground = cvzone.overlayPNG(imgBackground, imgWasteList[classID - 1], (909, 127))
        imgBackground = cvzone.overlayPNG(imgBackground, imgArrow, (978, 320))
        classIDBin = classDic[classID]

    if classIDBin is not None:
        imgBackground = cvzone.overlayPNG(imgBackground, imgBinsList[classIDBin], (895, 374))

    imgBackground[148:148 + 340, 159:159 + 454] = imgResize

    cv2.imshow("Image", img)
    cv2.imshow("Output", imgBackground)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()

```
