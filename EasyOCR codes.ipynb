```python
import cv2
import matplotlib.pyplot as plt
```

```python
img=cv2.imread(r"C:\Users\anshu\OneDrive\Desktop\Car dataset (1)\Mahindra Scorpio\254.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB))
```

```python
bfilter = cv2.bilateralFilter(gray , 4 , 7 , 7)
edged = cv2.Canny(bfilter,30 ,300)
plt.imshow(cv2.cvtColor(edged, cv2.COLOR_BGR2RGB))
```

```python
import cv2
import imutils

keypoints=cv2.findContours(edged.copy(),cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
contours = imutils.grab_contours(keypoints)
contours = sorted(contours , key =cv2.contoursArea, reverse =True)[:10]
```
