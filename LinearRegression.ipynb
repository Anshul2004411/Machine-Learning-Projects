```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
#sample data
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([1,4,9,16,25])
```

```python
#create polynomial features (degree 2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

#Train the model
model = LinearRegression()
model.fit(X_poly,y)
```

```python
#Predict

y_pred = model.predict(X_poly)
print(y_pred)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Dataset
X = np.array([[6], [8], [10], [12], [14], [16]])
y = np.array([80, 120, 180, 260, 360, 480])

# Polynomial features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Train model
model = LinearRegression()
model.fit(X_poly, y)

# Predict price for a 15-inch pizza
price = model.predict(poly.transform([[15]]))
print("Predicted Price:", price[0])

# Plot
plt.scatter(X, y, color='red')
plt.plot(X, model.predict(X_poly), color='blue')
plt.xlabel("Pizza Size (inches)")
plt.ylabel("Price (₹)")
plt.title("Polynomial Regression - Pizza Price Prediction")
plt.show()
```

```python
x=df.drop('Price',axis=1)
y=df["Price"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
from sklearn import metrics
```

```python
score = metrics.r2_score(y_test,y_pred)
print(score)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
data = {
    'Size': [8, 10, 12, 14, 16],
    'Toppings': [1, 2, 2, 3, 4],
    'Cheese': [1, 1, 2, 2, 3],
    'Price': [150, 220, 300, 400, 550]
}

df = pd.DataFrame(data)

X = df[['Size', 'Toppings', 'Cheese']]
y = df['Price']

model = LinearRegression()
model.fit(X, y)

# Predict the price of a 15-inch pizza with 3 toppings and 2 cheese portions
prediction = model.predict([[15, 3, 2]])
print("Predicted Price:", prediction[0])
```
