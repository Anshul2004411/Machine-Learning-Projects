```python
import pandas as pd
one=pd.DataFrame({
    'id':[1,2,3,4,5,],
    'Name':['Alex','Any','Allen','Alice','Ayoung'],
    'Subject_id':["sub1","sub2","sub3","sub4","sub5"]})
one
```

```python
import pandas as pd
two=pd.DataFrame({
    'id':[1,2,3,14,15],
    'Name':['Billy','Any','Bran','Bryce','Betty'],
    'Subject_id':["sub1","sub2","sub3","sub4","sub5"]})
two
```

```python
pd.merge(one,two,on='id')
```

```python
pd.merge(one,two,on='id',how="left")
```

```python
pd.merge(one,two,on='id',how="right")
```

```python
pd.merge(one,two,on='id',how="outer")
```

```python
pd.merge(one,two,on=['id','Name'],how="inner")
```
